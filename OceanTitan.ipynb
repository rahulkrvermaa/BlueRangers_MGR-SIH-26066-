```python
# Prepare OSCAR currents on the target grid

current_u_list = []
current_v_list = []

for ds in current_list:
    u = ds["u"]
    v = ds["v"]

    # Rename coordinate variables only
    u = u.rename({"lat": "latitude", "lon": "longitude"})
    v = v.rename({"lat": "latitude", "lon": "longitude"})

    # Make latitude/longitude proper dimension coordinates
    u = u.assign_coords(
        latitude=("latitude", u["latitude"].values),
        longitude=("longitude", u["longitude"].values)
    )

    v = v.assign_coords(
        latitude=("latitude", v["latitude"].values),
        longitude=("longitude", v["longitude"].values)
    )

    # Change longitude from 0–360 to -180–180
    u = u.assign_coords(
        longitude=((u.longitude + 180) % 360) - 180
    ).sortby("longitude")

    v = v.assign_coords(
        longitude=((v.longitude + 180) % 360) - 180
    ).sortby("longitude")

    # Put dimensions in standard order
    u = u.transpose("time", "latitude", "longitude")
    v = v.transpose("time", "latitude", "longitude")

    # Interpolate directly to target grid
    u = u.interp(
        latitude=target_lat,
        longitude=target_lon
    )

    v = v.interp(
        latitude=target_lat,
        longitude=target_lon
    )

    current_u_list.append(u)
    current_v_list.append(v)

# Combine all 15 days
current_u_all = xr.concat(current_u_list, dim="time")
current_v_all = xr.concat(current_v_list, dim="time")

print("Current U shape:", current_u_all.shape)
print("Current V shape:", current_v_all.shape)
print("Current U dimensions:", current_u_all.dims)
print("Current V dimensions:", current_v_all.dims)
```

```python
# Prepare CCMP wind U and V on the target grid

wind_u_list = []
wind_v_list = []

for ds in wind_list:
    u = ds["uwnd"]
    v = ds["vwnd"]

    # Rename coordinates
    u = u.rename({
        "latitude": "latitude",
        "longitude": "longitude"
    })

    v = v.rename({
        "latitude": "latitude",
        "longitude": "longitude"
    })

    # Select the regional area
    u = u.sel(
        latitude=slice(5, 30),
        longitude=slice(45, 105)
    )

    v = v.sel(
        latitude=slice(5, 30),
        longitude=slice(45, 105)
    )

    # Interpolate to target grid
    u = u.interp(
        latitude=target_lat,
        longitude=target_lon
    )

    v = v.interp(
        latitude=target_lat,
        longitude=target_lon
    )

    # Standard dimension order
    u = u.transpose("time", "latitude", "longitude")
    v = v.transpose("time", "latitude", "longitude")

    wind_u_list.append(u)
    wind_v_list.append(v)

# Combine all 15 days
wind_u_all = xr.concat(wind_u_list, dim="time")
wind_v_all = xr.concat(wind_v_list, dim="time")

print("Wind U shape:", wind_u_all.shape)
print("Wind V shape:", wind_v_all.shape)
print("Wind U dimensions:", wind_u_all.dims)
print("Wind V dimensions:", wind_v_all.dims)
```

```python
# After temporal still missing so  now we handle the remaining 5.54% using spatial nearest-neighbor filling.
import numpy as np
from scipy.spatial import cKDTree

sss_arr = sss_filled.values.copy()
mask = ocean_mask.values

lat2d, lon2d = np.meshgrid(
    sss_filled.latitude.values,
    sss_filled.longitude.values,
    indexing="ij"
)

for t in range(sss_filled.sizes["time"]):

    arr = sss_arr[t]

    valid = np.isfinite(arr) & mask
    missing = np.isnan(arr) & mask

    if valid.any() and missing.any():

        valid_idx = np.where(valid)
        missing_idx = np.where(missing)

        valid_points = np.column_stack(
            (lat2d[valid_idx], lon2d[valid_idx])
        )

        missing_points = np.column_stack(
            (lat2d[missing_idx], lon2d[missing_idx])
        )

        tree = cKDTree(valid_points)

        _, nearest = tree.query(missing_points)

        arr[missing_idx] = arr[
            valid_idx[0][nearest],
            valid_idx[1][nearest]
        ]

sss_spatial = xr.DataArray(
    sss_arr,
    dims=sss_filled.dims,
    coords=sss_filled.coords,
    name="sss"
)

sss_spatial = sss_spatial.where(ocean_mask)
```

### Train the model

```python
# Train the model and track training and validation loss

num_epochs = 30
patience = 5

train_losses = []
val_losses = []

best_val_loss = float("inf")
patience_counter = 0

for epoch in range(num_epochs):

    model.train()
    train_loss = 0.0

    for X_batch, Y_batch in train_loader:
        X_batch = X_batch.to(device)
        Y_batch = Y_batch.to(device)

        optimizer.zero_grad()

        predictions = model(X_batch)
        loss = criterion(predictions, Y_batch)

        loss.backward()
        optimizer.step()

        train_loss += loss.item() * X_batch.size(0)

    train_loss /= len(train_loader.dataset)

    model.eval()
    val_loss = 0.0

    with torch.no_grad():
        for X_batch, Y_batch in val_loader:
            X_batch = X_batch.to(device)
            Y_batch = Y_batch.to(device)

            predictions = model(X_batch)
            loss = criterion(predictions, Y_batch)

            val_loss += loss.item() * X_batch.size(0)

    val_loss /= len(val_loader.dataset)

    train_losses.append(train_loss)
    val_losses.append(val_loss)

    print(
        f"Epoch {epoch + 1:02d}/{num_epochs} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Val Loss: {val_loss:.4f}"
    )

    if val_loss < best_val_loss:
        best_val_loss = val_loss
        patience_counter = 0
        best_model_state = model.state_dict()
    else:
        patience_counter += 1

    if patience_counter >= patience:
        print("\nEarly stopping triggered.")
        break

model.load_state_dict(best_model_state)

print("\nBest validation loss:", best_val_loss)
```

```python
# Define masked MSE loss for CNN patches

def masked_mse_loss(predictions, targets, mask):
    mask = mask.unsqueeze(1)
    squared_error = (predictions - targets) ** 2
    masked_error = squared_error * mask

    return masked_error.sum() / mask.expand_as(squared_error).sum()


# Test the loss with one training batch

X_batch, Y_batch, mask_batch = next(iter(train_patch_loader))

test_predictions = torch.zeros_like(Y_batch)

test_loss = masked_mse_loss(
    test_predictions,
    Y_batch,
    mask_batch
)

print("Prediction shape:", test_predictions.shape)
print("Target shape:", Y_batch.shape)
print("Mask shape:", mask_batch.shape)
print("Test loss:", test_loss.item())
```

```python
# Reset CNN for patch-based training

model_cnn_patch = OceanTemperatureCNN().to(device)

criterion_cnn_patch = masked_mse_loss

optimizer_cnn_patch = torch.optim.Adam(
    model_cnn_patch.parameters(),
    lr=0.0005,
    weight_decay=1e-5
)

num_epochs_patch = 20
patience_patch = 5

train_losses_patch = []
val_losses_patch = []

best_val_loss_patch = float("inf")
patience_counter_patch = 0
best_model_state_patch = None

print("Patch CNN ready")
print("Device:", device)
print("Parameters:", sum(p.numel() for p in model_cnn_patch.parameters()))
```

```python
# Train the patch-based CNN

for epoch in range(num_epochs_patch):

    model_cnn_patch.train()
    running_train_loss = 0.0

    for X_batch, Y_batch, mask_batch in train_patch_loader:
        X_batch = X_batch.to(device)
        Y_batch = Y_batch.to(device)
        mask_batch = mask_batch.to(device)

        optimizer_cnn_patch.zero_grad()

        predictions = model_cnn_patch(X_batch)

        loss = criterion_cnn_patch(
            predictions,
            Y_batch,
            mask_batch
        )

        loss.backward()
        optimizer_cnn_patch.step()

        running_train_loss += loss.item()

    train_loss = running_train_loss / len(train_patch_loader)

    model_cnn_patch.eval()
    running_val_loss = 0.0

    with torch.no_grad():
        for X_batch, Y_batch, mask_batch in val_patch_loader:
            X_batch = X_batch.to(device)
            Y_batch = Y_batch.to(device)
            mask_batch = mask_batch.to(device)

            predictions = model_cnn_patch(X_batch)

            loss = criterion_cnn_patch(
                predictions,
                Y_batch,
                mask_batch
            )

            running_val_loss += loss.item()

    val_loss = running_val_loss / len(val_patch_loader)

    train_losses_patch.append(train_loss)
    val_losses_patch.append(val_loss)

    print(
        f"Epoch {epoch+1:02d}/{num_epochs_patch} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Val Loss: {val_loss:.4f}"
    )

    if val_loss < best_val_loss_patch:
        best_val_loss_patch = val_loss
        patience_counter_patch = 0
        best_model_state_patch = {
            key: value.cpu().clone()
            for key, value in model_cnn_patch.state_dict().items()
        }
    else:
        patience_counter_patch += 1

    if patience_counter_patch >= patience_patch:
        print("Early stopping triggered.")
        break

print("Best validation loss:", best_val_loss_patch)
```

```python
# Load the best patch CNN model

model_cnn_patch.load_state_dict(best_model_state_patch)
model_cnn_patch = model_cnn_patch.to(device)
model_cnn_patch.eval()

print("Loaded best patch CNN")
print("Best validation loss:", best_val_loss_patch)
```

```python
# Reconstruct ARGO-period patch predictions onto the full grid

n_argo_days = len(argo_model_dates)
n_depths = len(required_depths)

prediction_sum_argo = np.zeros(
    (n_argo_days, n_depths, 100, 240),
    dtype=np.float64
)

prediction_count_argo = np.zeros(
    (n_argo_days, 100, 240),
    dtype=np.float64
)

patch_index = 0

for t in range(n_argo_days):
    for i in range(n_lat_patches):
        for j in range(n_lon_patches):

            patch_mask = mask_np[
                i:i + patch_size,
                j:j + patch_size
            ]

            if patch_mask.mean() < min_ocean_fraction:
                continue

            prediction_patch = argo_predictions[patch_index]

            valid_mask = patch_mask.astype(bool)

            prediction_sum_argo[
                t,
                :,
                i:i + patch_size,
                j:j + patch_size
            ] += (
                prediction_patch *
                valid_mask[np.newaxis, :, :]
            )

            prediction_count_argo[
                t,
                i:i + patch_size,
                j:j + patch_size
            ] += valid_mask

            patch_index += 1

print("Patches reconstructed:", patch_index)
print("Expected patches:", n_argo_days * len(patch_masks))
```

```python
# Average overlapping patches and compute the June 5–14 period mean

reconstructed_argo = np.divide(
    prediction_sum_argo,
    prediction_count_argo[:, np.newaxis, :, :],
    out=np.full_like(prediction_sum_argo, np.nan),
    where=prediction_count_argo[:, np.newaxis, :, :] > 0
)

print("Daily reconstructed shape:", reconstructed_argo.shape)
print("Daily minimum:", float(np.nanmin(reconstructed_argo)))
print("Daily maximum:", float(np.nanmax(reconstructed_argo)))
print("Daily mean:", float(np.nanmean(reconstructed_argo)))

argo_period_prediction = np.nanmean(
    reconstructed_argo,
    axis=0
)

print("\nPeriod-mean prediction shape:", argo_period_prediction.shape)
print("Period minimum:", float(np.nanmin(argo_period_prediction)))
print("Period maximum:", float(np.nanmax(argo_period_prediction)))
print("Period mean:", float(np.nanmean(argo_period_prediction)))
```

```python
# Calculate independent OceanEmbed validation metrics against ARGO

argo_validation_results = []

for depth in common_depths:

    model_values = oceanembed_common.sel(depth=depth).values
    argo_values = argo_common.sel(ZAX=depth).values

    valid = (
        np.isfinite(model_values)
        & np.isfinite(argo_values)
    )

    model_valid = model_values[valid]
    argo_valid = argo_values[valid]

    if len(model_valid) > 1:

        error = model_valid - argo_valid

        rmse = np.sqrt(np.mean(error ** 2))
        mae = np.mean(np.abs(error))
        bias = np.mean(error)

        correlation = np.corrcoef(
            model_valid,
            argo_valid
        )[0, 1]

    else:
        rmse = np.nan
        mae = np.nan
        bias = np.nan
        correlation = np.nan

    argo_validation_results.append({
        "Depth (m)": depth,
        "RMSE (°C)": rmse,
        "MAE (°C)": mae,
        "Bias (°C)": bias,
        "Correlation": correlation,
        "Valid points": len(model_valid)
    })

argo_validation_df = pd.DataFrame(
    argo_validation_results
)

print(argo_validation_df.to_string(index=False))
```

```python
# Calculate overall independent ARGO validation metrics

all_model_values = []
all_argo_values = []

for depth in common_depths:

    model_values = oceanembed_common.sel(
        depth=depth
    ).values

    argo_values = argo_common.sel(
        ZAX=depth
    ).values

    valid = (
        np.isfinite(model_values)
        & np.isfinite(argo_values)
    )

    all_model_values.append(model_values[valid])
    all_argo_values.append(argo_values[valid])

all_model_values = np.concatenate(all_model_values)
all_argo_values = np.concatenate(all_argo_values)

overall_error = (
    all_model_values - all_argo_values
)

overall_rmse = np.sqrt(
    np.mean(overall_error ** 2)
)

overall_mae = np.mean(
    np.abs(overall_error)
)

overall_bias = np.mean(
    overall_error
)

overall_correlation = np.corrcoef(
    all_model_values,
    all_argo_values
)[0, 1]

print("Overall independent ARGO validation")
print("------------------------------------")
print(f"Valid observations: {len(all_model_values)}")
print(f"RMSE: {overall_rmse:.4f} °C")
print(f"MAE: {overall_mae:.4f} °C")
print(f"Bias: {overall_bias:.4f} °C")
print(f"Correlation: {overall_correlation:.4f}")
```

```python
# OceanEmbed interactive subsurface temperature predictor

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import ipywidgets as widgets
from IPython.display import display, clear_output


# --------------------------------------------------
# 1. Prepare dates and model inputs
# --------------------------------------------------

available_dates = pd.to_datetime(
    final_complete.time.values
)

features = [
    "sst",
    "sss",
    "ssh",
    "current_u",
    "current_v",
    "wind_u",
    "wind_v"
]


# --------------------------------------------------
# 2. Precompute scaled surface data
# --------------------------------------------------

surface_features = final_complete[
    features
].to_array(
    dim="feature"
).transpose(
    "time",
    "feature",
    "latitude",
    "longitude"
)

surface_scaled = (
    surface_features
    - checkpoint["train_mean"]
) / checkpoint["train_std"]

surface_scaled_values = (
    surface_scaled.values.astype(np.float32)
)

print(
    "Surface data prepared:",
    surface_scaled_values.shape
)


# --------------------------------------------------
# 3. Create widgets
# --------------------------------------------------

date_dropdown = widgets.Dropdown(
    options=[
        (
            d.strftime("%Y-%m-%d"),
            d
        )
        for d in available_dates
    ],
    value=available_dates[0],
    description="Date:"
)

lat_slider = widgets.FloatSlider(
    value=15.5,
    min=float(target_lat.min()),
    max=float(target_lat.max()),
    step=0.25,
    description="Latitude:",
    continuous_update=False
)

lon_slider = widgets.FloatSlider(
    value=65.5,
    min=float(target_lon.min()),
    max=float(target_lon.max()),
    step=0.25,
    description="Longitude:",
    continuous_update=False
)


# --------------------------------------------------
# 4. Prediction function
# --------------------------------------------------

def generate_profile(
    date,
    latitude,
    longitude
):

    clear_output(wait=True)

    # Find nearest grid point

    lat_idx = int(
        np.abs(
            target_lat - latitude
        ).argmin()
    )

    lon_idx = int(
        np.abs(
            target_lon - longitude
        ).argmin()
    )

    selected_lat = float(
        target_lat[lat_idx]
    )

    selected_lon = float(
        target_lon[lon_idx]
    )


    # Find selected date

    time_idx = int(
        np.abs(
            available_dates
            - pd.Timestamp(date)
        ).argmin()
    )


    # Get selected day's preprocessed data

    x_day = surface_scaled_values[
        time_idx
    ]


    # Patch size

    half = patch_size // 2


    # Keep patch inside model domain

    lat_idx = np.clip(
        lat_idx,
        half,
        len(target_lat) - half - 1
    )

    lon_idx = np.clip(
        lon_idx,
        half,
        len(target_lon) - half - 1
    )


    # Extract 15 x 15 surface patch

    x_patch = x_day[
        :,
        lat_idx-half:lat_idx+half+1,
        lon_idx-half:lon_idx+half+1
    ]


    # Convert patch to tensor

    x_patch_tensor = torch.tensor(
        x_patch,
        dtype=torch.float32
    ).unsqueeze(0).to(device)


    # Model prediction

    model_argo.eval()

    with torch.no_grad():

        prediction = model_argo(
            x_patch_tensor
        )


    prediction = (
        prediction
        .cpu()
        .numpy()[0]
    )


    # Extract center-point temperature profile

    depth_profile = prediction[
        :,
        half,
        half
    ]


    # --------------------------------------------------
    # 5. Display information
    # --------------------------------------------------

    print(
        "🌊 OceanEmbed Subsurface Temperature Prediction"
    )

    print(
        f"Date: {pd.Timestamp(date).strftime('%Y-%m-%d')}"
    )

    print(
        f"Location: "
        f"{selected_lat:.3f}°N, "
        f"{selected_lon:.3f}°E"
    )


    # --------------------------------------------------
    # 6. Display surface parameters
    # --------------------------------------------------

    surface_values = final_complete.sel(
        time=pd.Timestamp(date)
    ).isel(
        latitude=lat_idx,
        longitude=lon_idx
    )

    surface_table = pd.DataFrame({

        "Parameter": [
            "SST",
            "SSS",
            "SSH",
            "Current U",
            "Current V",
            "Wind U",
            "Wind V"
        ],

        "Value": [
            float(surface_values["sst"].values),
            float(surface_values["sss"].values),
            float(surface_values["ssh"].values),
            float(surface_values["current_u"].values),
            float(surface_values["current_v"].values),
            float(surface_values["wind_u"].values),
            float(surface_values["wind_v"].values)
        ],

        "Unit": [
            "°C",
            "PSU",
            "m",
            "m/s",
            "m/s",
            "m/s",
            "m/s"
        ]

    })

    print("\nSurface Ocean Conditions")

    display(surface_table.round(3))


    # --------------------------------------------------
    # 7. Plot temperature profile
    # --------------------------------------------------

    plt.figure(
        figsize=(7, 8)
    )

    plt.plot(
        depth_profile,
        required_depths,
        marker="o",
        linewidth=2
    )

    plt.gca().invert_yaxis()

    plt.xlabel(
        "Predicted Temperature (°C)",
        fontsize=12
    )

    plt.ylabel(
        "Depth (m)",
        fontsize=12
    )

    plt.title(
        "OceanEmbed Subsurface Temperature Profile",
        fontsize=14
    )

    plt.grid(
        True,
        alpha=0.3
    )

    plt.tight_layout()

    plt.show()


    # --------------------------------------------------
    # 8. Display prediction table
    # --------------------------------------------------

    profile_df = pd.DataFrame({

        "Depth (m)": required_depths,

        "Predicted Temperature (°C)": np.round(
            depth_profile,
            2
        )

    })

    print("\nPredicted Subsurface Temperature")

    display(profile_df)


# --------------------------------------------------
# 9. Connect widgets to prediction
# --------------------------------------------------

interactive_plot = widgets.interactive_output(
    generate_profile,
    {
        "date": date_dropdown,
        "latitude": lat_slider,
        "longitude": lon_slider
    }
)


# --------------------------------------------------
# 10. Display OceanEmbed dashboard
# --------------------------------------------------

display(
    widgets.VBox([

        widgets.HTML(
            """
            <h2>🌊 OceanEmbed</h2>
            <h4>
            AI-Based Subsurface Ocean Temperature Predictor
            </h4>
            """
        ),

        date_dropdown,

        lat_slider,

        lon_slider,

        interactive_plot

    ])
)
```
